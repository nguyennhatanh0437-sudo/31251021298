```python
#include <MD_Parola.h>
#include <MD_MAX72xx.h>
#include <SPI.h>

#define HARDWARE_TYPE MD_MAX72XX::FC16_HW
#define MAX_DEVICES 4

#define CLK_PIN   13
#define DATA_PIN  11
#define CS_PIN    10

MD_Parola myDisplay = MD_Parola(HARDWARE_TYPE, CS_PIN, MAX_DEVICES);

char msg[100] = "Ho Chi Minh City University of Economics      ";

const int RELAY_PIN = 5;
const int SENSOR_PIN = 6;

enum PumpMode { MODE_AUTO, MODE_MANUAL_ON, MODE_MANUAL_OFF };
PumpMode currentMode = MODE_AUTO;

void setup() {
  Serial.begin(9600);

  myDisplay.begin();
  myDisplay.setIntensity(1);
  myDisplay.displayClear();
  myDisplay.displayText(msg, PA_LEFT, 60, 0, PA_SCROLL_LEFT, PA_SCROLL_LEFT);

  pinMode(RELAY_PIN, OUTPUT);
  pinMode(SENSOR_PIN, INPUT);

  digitalWrite(RELAY_PIN, HIGH);
}

void loop() {
  if (myDisplay.displayAnimate()) {
    myDisplay.displayReset();
  }

  if (Serial.available() > 0) {
    String inputStr = Serial.readStringUntil('\n');
    inputStr.trim();

    if (inputStr == "P_AUTO") {
      currentMode = MODE_AUTO;
    }
    else if (inputStr == "P_ON") {
      currentMode = MODE_MANUAL_ON;
    }
    else if (inputStr == "P_OFF") {
      currentMode = MODE_MANUAL_OFF;
    }
    else if (inputStr.startsWith("T:")) {
      String newText = inputStr.substring(2) + "      ";
      newText.toCharArray(msg, 100);
      myDisplay.displayReset();
    }
  }

  if (currentMode == MODE_AUTO) {
    int water = digitalRead(SENSOR_PIN);

    if (water == LOW) {
      digitalWrite(RELAY_PIN, HIGH);
    } else {
      digitalWrite(RELAY_PIN, LOW);
    }
  }
  else if (currentMode == MODE_MANUAL_ON) {
    digitalWrite(RELAY_PIN, LOW);
  }
  else if (currentMode == MODE_MANUAL_OFF) {
    digitalWrite(RELAY_PIN, HIGH);
  }
}
```
