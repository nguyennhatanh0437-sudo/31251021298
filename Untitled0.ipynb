```python
import folium
from geopy.distance import geodesic
center = (10.783012628760616, 106.69519172424411)
m = folium.Map(location=center, zoom_start=14)
folium.Marker(center, popup="UEH ISB", icon=folium.Icon(color="red")).add_to(m)
locations = [
    ("Benh vien nhi dong 2", (10.782085160636296, 106.70227275590337)),
    ("Thao cam vien", (10.788998945127547, 106.70562015268776)),
    ("Van Hanh mall", (10.771039638722913, 106.66991458698767)),
    ("Benh vien da lieu", (10.776604608819204, 106.6866086555662)),
    ("Bao tang phu nu nam bo", (10.784277353413053, 106.68772445449433))
]
for name, coord in locations:
    dist = geodesic(center, coord).km
    folium.Marker(
        coord,
        popup=f"{name}<br>{dist:.2f} km",
        tooltip="Click"
    ).add_to(m)
    folium.PolyLine([center, coord], color="green").add_to(m)
m
```
